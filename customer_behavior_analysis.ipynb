now we got some null/missing values, we will replace these by median/mode, median is preferred over mean because it is robust to outliers while on the other hand mean is affected by the outliers which makes our data's distribution skewed. If we replace all the null values directly with the overall median of the given data, while it technically works it is not the best practise, why? because review ratings can vary a lot depending on the category, for example: clothing will have different review rating than footwear so if we take the overall median we are basically forcing the clothing category to have the same review rating as footwear category, this introduces bias into our dataset.In order to resolve this, what we will do is we will find median within each category and then replace the missing values in that category with the median of that category only. What this would do is that the missing review rating of a clothing item will be replaced by the median review rating of a clothing item only and not a footwear item. This decreases bias in our dataset.

```python
df['Review Rating']=df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

now our data is cleaned i.e. we don't have any null values now in any coloumn. We have respected the differences between category which gives us more accurate and trustworthy data.

now we observe our coloumn names, they have upper case, lower case letters and spaces in between, it will be difficult for an analyst to remember the exact format in large datasets so we use snake casing , it uses lower case letters and underscore. This makes our code consistent, easier to read and most importantly snake casing makes the referencing of our code in SQL or python easier without running into any errors.

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df= df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

doing some feature engineering and creating a new column age_group

```python
labels = ['Young Adult','Adult','Middle-aged','Senior']
df['age_group']=pd.qcut(df['age'], q=4, labels=labels)
```

```python
df[['age','age_group']].head(10)
```

we did all this to understand consumer behaviour patters across different age groups.

```python
# create column purchase_frequency_days

frequency_mapping = {
    'Fortnightly':14,
    'Weekly':7,
    'Monthly':30,
    'Quarterly':90,
    'Bi-Weekly':14,
    'Annually':365,
    'Every 3 Months':90
}
df['purchase_frequency_days']=df['frequency_of_purchases'].map(frequency_mapping)
```
